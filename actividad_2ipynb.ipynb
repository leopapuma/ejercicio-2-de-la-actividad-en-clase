para resolver el problema usuamos la libreria nnumpy

```python
import numpy as np
```

asignamos parametros para poder resolver el ejercicio

```python
# 1. Parametros
t = 1.0
N = 1000
M = 50000
dt = t / N
```

creamos el generador de numer aleatorios, usando la semilla 42

```python
# Generador reproducible
rng = np.random.default_rng(42)
```

```python
# Arreglo para almacenar una integral por trayectoria
resultados = np.empty(M)
```

crearemos la simulacion por bloques y aremos el movimiento browiano, y usaremos la suma ade riemann por la derecha

```python
# 2. Simulacion por bloques
tamano_bloque = 1000
for inicio in range(0, M, tamano_bloque):
  fin = min(inicio + tamano_bloque, M)
  m = fin- inicio
  # Incrementos independientes N(0, dt)
  dW = rng.normal(
      loc=0.0,
      scale=np.sqrt(dt),
      size=(m, N)
  )
  # Movimiento browniano: W_0 = 0
  W = np.cumsum(dW, axis=1)
  # Suma de Riemann por la derecha
  X = dt * np.sum(W**2, axis=1)
  resultados[inicio:fin] = X
```

simulamos la media y la varianza

```python
media_simulada = np.mean(resultados)
```

```python
varianza_simulada = np.var(resultados, ddof=1)
```

calculamos el valor teorico que es el valor real que da la media y la varianza para comparar con nuestro modelo

```python
# 4. Valores teoricos
media_teorica = t**2 / 2
varianza_teorica = t**4 / 3
```

calculamos los errores que tiene nuestra media simulada y varianza con los valores teoricos

```python
# 5. Errores absolutos
error_media = abs(media_simulada- media_teorica)
error_varianza = abs(varianza_simulada- varianza_teorica)
```

imprimimos los resultados que obtuvimos

```python
# 6. Presentacion de resultados
print("RESULTADOS DE LA SIMULACION")
print("-----------------------------------")
print(f"Numero de trayectorias: {M}")
print(f"Numero de pasos: {N}")
print(f"Delta t: {dt:.6f}")
print()
print(f"Esperanza teorica: {media_teorica:.6f}")
print(f"Esperanza simulada: {media_simulada:.6f}")
print(f"Error absoluto: {error_media:.6f}")
print()
print(f"Varianza teorica: {varianza_teorica:.6f}")
print(f"Varianza simulada: {varianza_simulada:.6f}")
print(f"Error absoluto: {error_varianza:.6f}")
```

se puede observar que la simulacion se acerca demasiado al resultado teorico obteniendo un errror absoluto de media de 0.002 y de varianza de 0.002, nuestro modelo puedo aproximarse adecuadamente al resultado teorico
